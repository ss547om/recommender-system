```python
df = pd.read_csv("contract_criteria_export.csv")
```

odstránenie cenových kritérií

```python
pattern = r'\bcena\b|\bcien\b|\bceny\b|\bnáklad'

#Filtrovanie riadkov, ktoré Neobsahujú tieto výrazy
df_necenove = df[~df['criterion'].str.contains(pattern, case=False, na=False)]
```

```python
df_necenove.to_csv("contract_criteria_necenove.csv", index=False, encoding="utf-8")
```
